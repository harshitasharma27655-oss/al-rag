```python
import os
from pypdf import PdfReader

# PDF auto pick
pdf_file = [f for f in os.listdir() if f.endswith(".pdf")][0]
print("Using file:", pdf_file)

# Read PDF
reader = PdfReader(pdf_file)

full_text = ""
for page in reader.pages:
    full_text += page.extract_text()

print("Text loaded ✅")
```

```python
def chunk_text(text, chunk_size=500):
    chunks = []
    for i in range(0, len(text), chunk_size):
        chunks.append(text[i:i+chunk_size])
    return chunks

chunks = chunk_text(full_text)

print("Chunks:", len(chunks))
```

```python
def search_chunks(query, chunks):
    results = []
    for chunk in chunks:
        if query.lower() in chunk.lower():
            results.append(chunk)
    return results[:3]
```

```python
def run_query(query):
    results = search_chunks(query, chunks)

    if len(results) == 0:
        print("Answer:\nI don’t have that information in the document.")
        return

    print("Answer:\n")
    print("\n".join(results[:2])[:500])

    print("\nWhy:\nBased on retrieved document chunks.")

    print("\nCitations:")
    for i in range(len(results)):
        print(f"Chunk {i+1}")

    print("\nAssumptions:\nBased only on provided PDF.")
```

```python
run_query("What is this assessment?")
run_query("What is deadline?")
run_query("Explain submission requirements")
run_query("What is assessment 1?")
```
